```python
df.drop_duplicates(inplace =True)
```

```python
df.isnull().sum()
```

```python

df['Employment_Type'] = df['Employment_Type'].fillna(df['Employment_Type'].mode()[0])
df['Payment_History'] = df['Payment_History'].fillna(df['Payment_History'].mode()[0])

```

```python
df['Credit_Score'] = df['Credit_Score'].fillna(df['Credit_Score'].median())
df['Annual_Income'] = df['Annual_Income'].fillna(
    df.groupby('Employment_Type')['Annual_Income'].transform('median')
)
```
